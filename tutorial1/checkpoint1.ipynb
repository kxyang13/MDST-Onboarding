<h2> Feature Engineering </h2>

After exploring the data, you can choose to modify features that you would use to predict the performance of the students on your chosen response variable. 

You can also create your own features. For example, perhaps you figured that maybe a state's expenditure per student may affect their overall academic performance so you create a expenditure_per_student feature.

Use this space to modify or create features.

```python
# @@@ 5
df["RELATIVE_INSTRUCTION_EXPENDITURE"] = df["INSTRUCTION_EXPENDITURE"] / df["TOTAL_EXPENDITURE"] * 100
df["EXPENDITURE_PER_STUDENT"] = df["INSTRUCTION_EXPENDITURE"] / df["GRADES_ALL_G"]
df.head()
```

Feature engineering justification: **<I added the column "Relative_Instruction_Expenditure" to get a  rough estimate of how much each state prioritized spending on education as a percentage of total expenditure each year. <h2> Data Creation </h2>

_Use this space to create train/test data_

```python
from sklearn.model_selection import train_test_split
```

```python
# @@@ 8
# Setting avg. instruction expenditure/student and average 4th grade math scores as inputs 
X = df[["EXPENDITURE_PER_STUDENT", "AVG_MATH_4_SCORE"]].dropna()
y = df.loc[X.index]['AVG_READING_4_SCORE']

#filling missing y values with median
y.fillna(y.median(), inplace=True)
```

```python
# @@@ 9 

#setting test size to 20% and ensuring consistent results by setting random_state
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=0)
```

<h2> Prediction </h2>

ML Models [Resource](https://medium.com/@vijaya.beeravalli/comparison-of-machine-learning-classification-models-for-credit-card-default-data-c3cf805c9a5a)

```python
# @@@ 10
# import your sklearn class here
from sklearn.linear_model import LinearRegression
```

```python
# @@@ 11
# create your model here
model = LinearRegression() 
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

## Evaluation

Choose some metrics to evaluate the performance of your model, some of them are mentioned in the tutorial.

```python
# @@@ 12
#testing for R^2 value, mean error, mean absolute error, and root mean squared error
model.score(X_test, y_test)
np.mean(model.predict(X_test)-y_test)
np.mean(np.abs(model.predict(X_test)-y_test))
np.mean((model.predict(X_test)-y_test)**2)**0.5
```
